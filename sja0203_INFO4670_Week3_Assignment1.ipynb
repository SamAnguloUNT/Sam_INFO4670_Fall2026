### Q4 · Two quantitative variables
**Is there a relationship between how far students commute and their final GPA?** Remember: *association is not causation.*
→ *scatterplot with a trend line*. Columns: `commute_miles` and `final_gpa` (from `students`).

```python
fig, ax = plt.subplots()
sns.regplot(data=students, x="commute_miles", y="final_gpa",
            scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4},
            line_kws={"color": RUST}, ax=ax)
ax.set_title("Commute Distance vs. Final GPA")
ax.set_xlabel("Commute Distance (miles)")
ax.set_ylabel("Final GPA")
plt.show()
```

**STAR story — Q4** *(write in your own words)*

- **Scope —** This scatterplot shows the relationship between Final GPA which is the Y axis and the commute distance students travel to school in the X axis. This plot will show if a students commute distance affects their final GPA.


- **Track —** We are answering the question of whether the time it takes for a student to commute to school, affect their final gpa?


- **Articulate —** The chart shows that most students commute is between 0 to 10 miles with some outliers having a commute distance of 60 to 80 miles. Even with a low commute distance most students final GPA's are around 2 to 3. There are more students with a GPA above 3 when their commute distance is from 0 to 10 miles.

### Q5 · Over time
**How does student engagement change across the semester?**
→ *line chart* of average weekly activity. Use the `weekly` table: `week` and `minutes_active` (you'll need to average by week first — see the Guided notebook).

```python
weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()
fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active",
             color=NAVY, marker="o", ax=ax)
ax.set_title("Average active minutes per student, by week")
ax.set_xlabel("Week of semester")
ax.set_ylabel("Avg. minutes active")
plt.show()
```

**STAR story — Q5** *(write in your own words)*

- **Scope —** The Line Chart on the X axis shows the average minutes that student is active doing work on Canvas or whatever they use, some schools use Blackboard. The Y axis shows what week in the semester the students are in.

- **Track —** The question is answering how active students are doing work during each week of the semester.

- **Articulate —** It shows that students are doing about 200 minutes of work every week on Canvas, but during week 9 it plummets all the way down. The reason is because usually week 9 is some sort of week break where students have a whole week off of school.

- **Respond —** I notice that in week 9 the minutes active go all the way to zero, while it isn't unexpected for less work to be done in week 9, I dont think it would be zero.
